```python
import folium
hualien_center = [23.9769, 121.6044]
wood_shops = [
    {"name": "大源木業股份有限公司", "lat": 23.9631485, "lon": 121.5733806, "year": 1969},
    {"name": "朝盛木材行", "lat": 23.9765539, "lon": 121.586245, "year": 2014},
    {"name": "大永發建材行", "lat": 23.9712618, "lon": 121.6014286, "year": 1979},
    {"name": "東新木業行", "lat": 23.9721847, "lon": 121.5780551, "year": 1980},
    {"name": "嘉益木材行", "lat": 23.9511861, "lon": 121.590686, "year": 1973},
    {"name": "松梧木材行", "lat": 23.95916, "lon": 121.593787, "year": 1994},
    {"name": "勝裕建材行", "lat": 23.9977314, "lon": 121.6017339},
    {"name": "華信昌實業", "lat": 23.9547848, "lon": 121.5631098, "year": 2004},
    {"name": "崇成建材行", "lat": 23.9736972, "lon": 121.5737778, "year": 2005},
    {"name": "老傳建材行", "lat": 23.9728321, "lon": 121.5996267, "year": 2002},
    {"name": "豪臆建材行", "lat": 23.9829324, "lon": 121.5812487, "year": 1992},
    {"name": "祥捷木業", "lat": 23.9751049, "lon": 121.5736817, "year": 1992},
    {"name": "立健木業", "lat": 23.9805185, "lon": 121.5898376, "year": 1993},
    {"name": "吉興板模", "lat": 23.9779513, "lon": 121.5888936, "year": 1977},
    {"name": "宏銘企業股份有限公司", "lat": 23.9764884, "lon": 121.5848797, "year": 1967},
    {"name": "力元建材行", "lat": 23.952665, "lon": 121.590732, "year": 1983},
    {"name": "一成木材行", "lat": 23.9729168, "lon": 121.6015107, "year": 1978},
    {"name": "燕茂企業股份有限公司", "lat": 23.9571253, "lon": 121.593387, "year": 1973},
]

def year_category(year):
    if year is None:
        return "無年份"
    elif year < 1990:
        return "1990年以前"
    else:
        return "1990年（含）以後"
category_colors = {
    "1990年以前": "green",
    "1990年（含）以後": "red",
    "無年份": "gray"
}

m = folium.Map(
    location=hualien_center,
    zoom_start=13,
    tiles="CartoDB positron",
    control_scale=True 
)

for shop in wood_shops:
    year = shop.get("year")
    category = year_category(year)
    color = category_colors[category]

    popup_text = shop["name"]
    popup_text += f"<br>營業登記年：{year}" if year else "<br>營業登記年：未知"

    folium.Marker(
        location=[shop["lat"], shop["lon"]],
        popup=popup_text,
        icon=folium.Icon(color=color, icon="tree", prefix="fa")
    ).add_to(m)

north_arrow = """
<div style="
    position: fixed;top: 20px;right: 20px;width: 60px;height: 60px;z-index:9999;font-size: 18px;text-align: center;font-weight: bold;color: black;">
    ↑<br>N
</div>
"""
m.get_root().html.add_child(folium.Element(north_arrow))
m
```
